```python
N=10
k=5

R=N
c=1

while c != 0:
    d = int(input("no of candies:"))

    if d > R :
      print("INVALID INPUT")
      break

    R = R - d
    print("no of candies sold:",d)

    if R <= k:
        R = N

    print("no of candies available:", R)

    c = int(input("enter 1 to continue, 0 to stop: "))
```

```python
n = int(input("enter the no of students: "))

ranks = input().split()

for i in range(n):
     ranks[i] = int(ranks[i])

count = 0

for i in range(1, n):
    if ranks[i] < ranks[i - 1]:
        count += 1

print(count)
```

```python

r = int(input("enter the no of rats: "))
unit = int(input("enter the unit: "))
n = int(input("enter the number of food: "))

arr = [int(x) for x in input("arr: ").split()]

if not arr:
    print(-1)
else:
    required_food = r * unit
    total = 0
    houses = 0

    for food in arr:
        total += food
        houses += 1

        if total >= required_food:
            break

    if total < required_food:
        print(0)
    else:
        print(houses)
```
